```python
A = [-3, -1, 0, 1, 4, 7]

# Naive O(n) searching
if 8 in A:
    print("true")

```

```python
# traditional binary search - Looking up if number is in array:
# Time Comlexity; O(logn)
# space Complexity: O(1)

def binary_search(arr, target):
    N = len(arr)
    left, right = 0, N - 1

    while left <= right:
        mid = left + (right - left) // 2

        if arr[mid] == target:
            return True
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return False

binary_search(A, 4)
```

```python
# Condition based binary search - Finding first occurrence of a number in array


B = [False, False, False, True, True, True, True]

def binary_search_condition(arr):
    N = len(arr)
    left, right = 0, N - 1
    result = -1

    while left <= right:
        mid = left + (right - left) // 2

        if arr[mid]:  # Condition is met
            result = mid
            right = mid - 1  # Look for earlier occurrence
        else:
            left = mid + 1

    return result

binary_search_condition(B)
```
