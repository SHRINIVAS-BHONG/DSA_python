```python
n = 8
A = [[0,1], [1,2], [0,3], [3,4], [3,6], [4,2], [4,5], [5,2]]

A
```

```python
# Convert Array of Edges to Adjacency List

M = []
for i in range(n):
    M.append([0]*n)

for u, v in A:
    M[u][v] = 1

    #Uncommon The following line if graph is undirected
    #M[v][u] = 1

M
```

```python
# convert Array of Edges -> Adjacency List
from collections import defaultdict

D = defaultdict(list)

for u, v in A:
    D[u].append(v)

D
```

```python
# DFS with recursion - o(V + E) where V is the numberof nodes and E is the number of edges


def dfs_recursive(node):
    print(node)
    for neil_mode in D[node]:
        if neil_mode not in seen:
            seen.add(neil_mode)
            dfs_recursive(neil_mode)

source = 0
seen = set()
seen.add(source)
dfs_recursive(source)
```

```python
# iterative DFS - o(V + E) where V is the numberof nodes and E is the number of edges

source = 0
seen = set()
stack = [source]

while stack:
    node = stack.pop()
    print(node)
    for neil_mode in D[node]:
        if neil_mode not in seen:
            seen.add(neil_mode)
            stack.append(neil_mode)

```

```python
# BFS - o(V + E) where V is the numberof nodes and E is the number of edges

source = 0

from collections import deque

seen = set()
seen.add(source)
q = deque([source])

while q:
    node = q.popleft()
    print(node)
    for neil_mode in D[node]:
        if neil_mode not in seen:
            seen.add(neil_mode)
            q.append(neil_mode)
```

```python
class node:
    def __init__(self, val):
        self.val = val
        self.neighbors = []
    
    def __str__(self):
        return str(self.val)
    
    def display(self):
        connections = [node.value for node in self.neighbors]
        return f"Node({self.val}) -> {connections}"
    

A = node('A')
B = node('B')
C = node('C')
D = node('D')

A.neighbors.append(B)
A.neighbors.append(C)

C.neighbors.append(D)   
D.neighbors.append(C)

```
