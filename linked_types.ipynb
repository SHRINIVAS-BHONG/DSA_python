# singly linked list

```python
class SinglyNode:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next
    
    def __str__(self):
        return str(self.data)
```

```python
head = SinglyNode(10)
A = SinglyNode(20)
B = SinglyNode(30)
C = SinglyNode(40)

head.next = A
A.next = B
B.next = C

print(head)
```

# Traverse the linkedlist - O(n)

```python
curr = head

while curr is not None:
    print(curr)
    curr = curr.next
```

# Display Linked List

```python
def display(head):
    curr = head
    elements = []
    while curr is not None:
        elements.append(str(curr))
        curr = curr.next
    print(" -> ".join(elements))

display(head)
```

# Search for Node

```python
def search(head, key):
    curr = head
    while curr is not None:
        if curr.data == key:
            return True
        curr = curr.next
    return False

search(head, 30)
```

# doubly LinkedList

```python
class doublyNode:
    def __init__(self, data, prev=None, next=None):
        self.data = data
        self.prev = prev
        self.next = next
    
    def __str__(self):
        return str(self.data)
```

```python
head = tail = doublyNode(10)
A = doublyNode(20)
B = doublyNode(30)
C = doublyNode(40)

for node in (A, B, C):
    tail.next = node
    node.prev = tail
    tail = node

print(head)
print(tail)
```

```python
def display(head):
    curr = head
    elements = []
    while curr is not None:
        elements.append(str(curr))
        curr = curr.next
    print(" <-> ".join(elements))

display(head)
```

```python
def insert_at_beginning(head, tail, data):
    new_node = doublyNode(data)
    head.prev = new_node
    return new_node, tail

head, tail = insert_at_beginning(head, tail, 5)
display(head)
display(tail)
```
