```python
# Binary Tree
class TreeNode:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
    
    def __str__(self):
        return f"TreeNode({self.val})"
```

```python
#       1
#    2     3
# 4    5  10
```

```python
A = TreeNode(1)
B = TreeNode(2)
C = TreeNode(3)
D = TreeNode(4)
E = TreeNode(5)
F = TreeNode(10)

A.left = B
A.right = C
B.left = D
B.right = E
C.right = F

print(A)  # Output: TreeNode(1)
```

```python
# Recursive Preorder Traversal time: O(n), space: O(n)
def pre_order(node):
    if not node:
        return

    print(node.val)
    pre_order(node.left)
    pre_order(node.right)

pre_order(A)
```

```python
# recursive Inorder Traversal time: O(n), space: O(n)
def in_order(node):
    if not node:
        return

    in_order(node.left)
    print(node.val)
    in_order(node.right)

in_order(A)
```

```python
# recursive post-order Traversal time: O(n), space: O(n)
def post_order(node):
    if not node:
        return

    post_order(node.left)
    post_order(node.right)
    print(node.val)

post_order(A)
```

```python
# Iterative pre-order Traversal time: O(n), space: O(n)
def iterative_pre_order(root):
    stk = [root]
    
    while stk:
        node = stk.pop()
        if node:
            print(node.val)
            stk.append(node.right)
            stk.append(node.left)
    
iterative_pre_order(A)
```

```python
from collections import deque

def level_order(root):
    queue = deque([root]) 
    
    while queue:
        node = queue.popleft()
        if node:
            print(node.val)
            queue.append(node.left)
            queue.append(node.right)

level_order(A)
```

```python
# check if value exists time: O(n), space: O(n)

def search_value(root, target):
    if not root:
        return False
    
    if root.val == target:
        return True
    
    return search_value(root.left, target) or search_value(root.right, target)

print(search_value(A, 10))  # Output: True
print(search_value(A, 7))   # Output: False
```

```python
# Binary search tree(BSTs)

#        5
#     1      8
#  -1   3  7  9

A2 = TreeNode(5)
B2 = TreeNode(1)
C2 = TreeNode(8)
D2 = TreeNode(-1)
E2 = TreeNode(3)
F2 = TreeNode(7)
G2 = TreeNode(9)

A2.left, A2.right = B2, C2
B2.left, B2.right = D2, E2
C2.left, C2.right = F2, G2

print(A2)
```

```python
in_order(A2)
```

```python
# Time: O(log n), Space: O(log n)
def search_bst(root, target):
    if not root:
        return False
    
    if root.val == target:
        return True
    elif target < root.val:
        return search_bst(root.left, target)
    else:
        return search_bst(root.right, target)
    
print(search_bst(A2, 7))  # Output: True
```
