# Stack and Queue

```python
stk = []
print(stk)
```

```python
stk.append(5)
stk.append(10)
stk.append(15)

stk
```

```python
# pop form stack - O(1)
x = stk.pop()

print(x)
print(stk)
```

```python
print(stk[-1])
```

```python
if stk:
    print("Stack is not empty")
```

```python
stk.pop()
```

```python
# queues - first in first out

from collections import deque

q = deque()
print(q)
```

```python
# enqueue - O(1)
q.append(5)
q.append(10)

q
```

```python
#Dequeue (pop left) - Remove element from the left - O(1)

q.popleft()  # dequeue - O(1)
```

```python
# peek from left side - O(1)
q[0]
```

```python
# peek form right side - O(1)
q[-1]
```
