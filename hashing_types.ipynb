# hashing


```python
# hashset

s = set()
print(s)
```

```python
s.add(10)
s.add(20)
s.add(30)

print(s)
```

```python
if 10 not in s:
    print("10 not found")
```

```python
s.remove(20)
print(s)
```

```python
string = "aaaaaanbbbhhhheeeeeiiii"
sett = set(string)

print(sett)
```

```python
# loop over items in set - O(n) time

for items in sett:
    print(items)
```

```python
# hashmaps - Dictonaries
d = {'greg': 25, 'alice': 30, 'bob': 22}
print(d)
```

```python
d['alice'] = 31

print(d)
```

```python
# check for presence of key in dictionay: O(1) time

if 'greg' in d:
    print("greg is found")
```

```python
# check the value corresponding to a key in the dictionary: O(1) time
print(d['bob'])
```

```python
# loop ober the key:val pairs of the dictionary: O(n) time
for key, val in d.items():
    print(f"{key}: {val}")
```
