# Recursion

```python
# Fibonacci 
# F(0) = 0
# F(1) = 1 
# n > 1 : F(n) = F(n-1) + F(n-2)

# Time complexity: O(2^n)
#
def F(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return F(n-1) + F(n-2)

F(4)
```

```python
# singly linked list

class SinglyNode:
    def __init__(self, data):
        self.data = data
        self.next = None
    
    def __str__(self):
        return str(self.data)
    
head = SinglyNode(1)
A = SinglyNode(2)
B = SinglyNode(3)
C = SinglyNode(4)

head.next = A
A.next = B
B.next = C

print(head)
```

```python
# linkedlist

# Time complexity: O(n)
def reverse(node):
    if not node:
        return 
    
    reverse(node.next)
    print(node)

reverse(head)
```
